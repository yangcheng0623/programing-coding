##計算費伯納西數列第91個元素
1,1,2,3,5,8.....
請問第91個元素為何?

```python
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
fib[2]=fib[0]+fib[1]
print(fib)
fib[3]=fib[1]+fib[2]
print(fib)
fib[4]=fib[2]+fib[3]
print(fib)
```

```python
for i in range(2,91):
    fib[i]=fib[i-2]+fib[i-1]
print(fib[90])
```
